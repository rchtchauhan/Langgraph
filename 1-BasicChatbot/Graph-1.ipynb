```python
from typing import TypedDict
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    message : str

def greeting(state:AgentState)->AgentState:
    '''Simple node that add greeting '''
    state['message'] = 'hey ' + state['message'] + ' how are u doing'
    return state

def complement(state:AgentState)->AgentState:
    '''Node that compliment the user '''
    state['message'] = "hey " + state['message']+ ' you are doing a fantastic job buddy'
    return state
```

```python
graph = StateGraph(AgentState)
graph.add_node("greet",greeting)

graph.set_entry_point("greet")
graph.set_finish_point("greet")

app = graph.compile()

```

```python
result = app.invoke({"message":"Rachit"})   
result["message"]
```
