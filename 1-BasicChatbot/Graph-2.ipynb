### Graph 2


```python
from langgraph.graph import StateGraph
from typing import TypedDict, List
```

```python
class AgentState(TypedDict):
    nums : List[int]
    name : str
    operation : str
    result : str
```

```python
def calculation(state:AgentState)->AgentState:
    if state['operation'] == "+":
        state['result'] = f"Hey  {state['name']} 'your answer is {sum(state['nums'])}"
    else:
        k = 1
        for i in state['nums']:
            k = k * int(i)
        state['result'] = f" Hey  {state['name']}  'your answer is {k}" 
    return state
```

```python
graph = StateGraph(AgentState)
graph.add_node("calculate",calculation)
graph.set_entry_point("calculate")
graph.set_finish_point("calculate")
```

```python
app = graph.compile()
```

```python
result = app.invoke({"nums" : [4,1,6,9],
                "name" : 'Rachit',
                "operation": "*"})
print(result)
```
