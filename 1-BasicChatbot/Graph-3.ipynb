```python
from langgraph.graph import StateGraph
from typing import List, TypedDict
```

```python
class AgentState(TypedDict):
    name : str
    age : int
    skills : List
    result : str
```

```python
def first_greeting_node(state:AgentState)->AgentState:
    '''this node with do greeting'''
    state['result'] = f'hey {state['name']} how are u? '
    return state

def second_age_node(state:AgentState)->AgentState:
    state['result'] = state['result'] + f'You are {state["age"]} years old '
    return state

def third_skills_node(state:AgentState)->AgentState:
    state['result'] += f" your skills are {', '.join(state['skills'])}"    
    return state
```

```python
graph = StateGraph(AgentState)
graph.add_node("greet",first_greeting_node)
graph.add_node("age",second_age_node)
graph.add_node("skills",third_skills_node)

graph.set_entry_point("greet")
graph.set_finish_point("skills")
graph.add_edge("greet","age")
graph.add_edge("age","skills")

add = graph.compile()
```

```python
result = add.invoke({
    "name":"Rachit",
    'age': 23,
    'skills': ["pandas","numpy",'django','fastapi']
})
print(result)
```
